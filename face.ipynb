```python
# draw label on camera 
def draw_label(img,text,pos,bg_color):
    text_size = cv2.getTextSize(text,cv2,FONT_HERSHEY_SIMPLEX,1,cv2.FILLED)
    end_x = pos[0]+text_size[0][0] +2
    end_y = pos[0]+text_size[0][0] -2

    cv2.rectangle(img,pos,(end_x,end_y),bg_color,cv2.FILLED)
    cv2.putText(img,text,pos,cv2.FONT_HERSHEY_SIMPLEX,1,(0,0,0),1,cv2.LINE_AA)
```

```python
import cv2
import numpy as np
from keras.applications.vgg16 import preprocess_input

# Make sure your video capture object is initialized
# cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame.")
        break

    # 1. Detect faces using your existing function
    coods = detect_face(cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY))

    for x, y, w, h in coods:
        # 2. Crop the face area out of the original color frame
        face_roi = frame[y : y + h, x : x + w]

        # Safety check to ensure the crop contains valid pixels
        if face_roi.size == 0:
            continue

        # 3. Preprocess the cropped face exactly for VGG16
        face_rgb = cv2.cvtColor(face_roi, cv2.COLOR_BGR2RGB)
        face_resized = cv2.resize(face_rgb, (224, 224))
        face_preprocessed = preprocess_input(face_resized)

        # 4. Get the mask prediction (re-shaping to add batch dimension)
        prediction = model.predict(
            face_preprocessed.reshape(1, 224, 224, 3), verbose=0
        )

        # 5. Set labels and box colors based on the prediction threshold
        # (Swap the condition if your class labels are reversed)
        if prediction[0][0] < 0.5:
            label = "Mask"
            color = (0, 255, 0)  # Green in BGR
        else:
            label = "No Mask"
            color = (0, 0, 255)  # Red in BGR

        # 6. Draw the rectangle around the face
        cv2.rectangle(frame, (x, y), (x + w, y + h), color, 2)

        # 7. Draw the text label right above the rectangle
        cv2.putText(
            frame,
            label,
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            color,
            2,
            cv2.LINE_AA,
        )

    # Display the resulting frame
    cv2.imshow("Face Mask Detector", frame)

    # Press 'x' to exit the loop
    if cv2.waitKey(1) & 0xFF == ord("x"):
        break

cap.release()
cv2.destroyAllWindows()
```
